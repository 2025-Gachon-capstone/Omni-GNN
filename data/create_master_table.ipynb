```python
import pandas as pd

# 데이터 로딩
orders = pd.read_csv("orders.csv")
prior = pd.read_csv("order_products__prior.csv")
train = pd.read_csv("order_products__train.csv")
products = pd.read_csv("products.csv")
aisles = pd.read_csv("aisles.csv")
departments = pd.read_csv("departments.csv")
```

## 데이터 통합

```python
# prior, train 주문 통합
order_products = pd.concat([prior, train], axis=0)
```

```python
# 주문 정보 병합: order_id 기준
orders_sub = orders[["order_id", "user_id", "order_number", "order_hour_of_day", "days_since_prior_order", "eval_set"]]
df = order_products.merge(orders_sub, on="order_id", how="left")
df
```

```python
# 상품 → aisle, department 정보 추가
df = df.merge(products, on="product_id", how="left")
df
```

```python
# 상품 → aisle, department 정보 추가
df = df.merge(aisles, on="aisle_id", how="left")
df
```

```python
# 상품 → aisle, department 정보 추가
df = df.merge(departments, on="department_id", how="left")
df
```

## df 정렬 및 분석

```python
df = df[[
    "user_id",
    "eval_set",

    "order_id",
    "order_number",
    "order_hour_of_day",
    "days_since_prior_order",

    "product_id",
    "product_name",
    "add_to_cart_order",
    "reordered",

    "aisle_id",
    "aisle",
    "department_id",
    "department"
]]

df.head()
```

```python
# test는 order_products_prior과 train에 없었기 대문에 병합시 자동 삭제됨 
df[df['eval_set']=='test'].head()
```

```python
df = df.sort_values(by=["user_id", "order_number", "add_to_cart_order"]).reset_index(drop=True)
df.info()
```

```python
df.isnull().sum()
```

## days_since_prior_order는 첫주문시 None이 맞음

```python
# 'days_since_prior_order' 컬럼이 NaN인 행만 필터링
nan_days_since_prior_order_rows = df[df['days_since_prior_order'].isna()]

# 필터링된 행들의 'order_number' 값 확인
# 고유한 order_number 값들을 출력하여 1 외의 값이 있는지 확인합니다.
unique_order_numbers_with_nan = nan_days_since_prior_order_rows['order_number'].unique()

unique_order_numbers_with_nan
```

## test 라벨링

```python
# 1. 유저별 마지막 order_number 계산
user_max_order = df.groupby("user_id")["order_number"].max().reset_index()
user_max_order.columns = ["user_id", "max_order_number"]

# 2. 병합해서 각 row가 마지막 주문인지 표시
df = df.merge(user_max_order, on="user_id", how="left")
df["is_last_order"] = df["order_number"] == df["max_order_number"]

# 3. 마지막 주문이면서 eval_set이 prior이면 → test로 변경
df.loc[(df["eval_set"] == "prior") & (df["is_last_order"]), "eval_set"] = "test"

# 4. 중간 컬럼 제거
df.drop(columns=["max_order_number"], inplace=True)

df
```
